## 📖 Project Overview

Noise contamination is a universal challenge in signal processing, affecting data integrity in fields ranging from biomedical engineering to astrophysics. [cite_start]Existing tools are often "domain-specific," relying on fixed noise models that fail when applied to novel datasets[cite: 15, 25].

This project presents a **General-Purpose Fourier-Domain Framework** designed to democratize signal analysis. [cite_start]By leveraging the **Fast Fourier Transform (FFT)** and a modular filter design engine, this system provides a mathematically rigorous and computationally efficient tool for analyzing and denoising diverse one-dimensional datasets[cite: 16, 17].

### Key Capabilities
* [cite_start]**Domain Agnostic:** Proven effectiveness on ECG (Biomedical), P-waves (Seismic), and Audio signals[cite: 19, 42].
* [cite_start]**Spectral Precision:** Manipulates signals in the frequency domain using $O(N \log N)$ FFT algorithms[cite: 53].
* [cite_start]**Quantitative Validation:** Built-in calculation of **SNR** (Signal-to-Noise Ratio), **MSE** (Mean Squared Error), and **PSD** (Power Spectral Density)[cite: 18].

---

## 🏗️ System Architecture

[cite_start]The framework is modularized into four distinct processing units[cite: 55]:

1.  [cite_start]**Preprocessing Module:** Handles normalization, detrending, and windowing (Hann/Hamming) to mitigate spectral leakage[cite: 56].
2.  [cite_start]**Spectral Transformation Core:** Executes FFT/IFFT operations and computes magnitude/phase spectra[cite: 57].
3.  [cite_start]**Filter Design Unit:** A configurable engine for creating digital filters[cite: 58]:
    * *Low-Pass / High-Pass* (Butterworth & Ideal)
    * *Band-Pass / Band-Stop*
    * *Notch Filters* (Specific frequency rejection, e.g., 60Hz mains hum)
    * *Adaptive Spectral Subtraction*
4.  [cite_start]**Reconstruction & Evaluation:** Rebuilds time-domain signals and computes error metrics[cite: 59].
